```python
catalog = pd.read_csv('catalog.csv')
catalog
```

```python
transactions = pd.read_csv('transactions.csv')
transactions
```

```python
warranty_map = {
    '2 years': 24,
    '1 year': 12,
    '3 years': 36,
    '6 months': 6
}

catalog['warranty_months'] = catalog['warranty'].map(warranty_map)
```

```python
for cc in ['product', 'color', 'size', 'category', 'brand', 'material', 'warranty']:
    print(cc)
    print(catalog[cc].nunique())
    print(catalog[cc].min())
    print(catalog[cc].max())
    grouped = catalog.groupby(cc)['product'].count()
    print(grouped.describe())
    print('')
```

```python
for nn in ['price', 'weight', 'quantity', 'rating', 'warranty_months']:
    print(nn)
    print(catalog[nn].describe())
```
